## 6.1 RGB color space

```python
#Answer your code here.
dog = cv2.imread('./Dogg.jpg')
image_rgb = dog.copy()
image_rgb  = cv2.cvtColor(image_rgb, cv2.COLOR_BGR2RGB)

# linear function: f(x)=ax+b
R, G, B = cv2.split(image_rgb)

# Apply linear transformations
a = 1.2  # Multiplication constant (scaling factor)
b = 30   # Addition constant (offset)

R_transformed = np.clip(a * R + b, 0, 255).astype(np.uint8)
G_transformed = np.clip(a * G + b, 0, 255).astype(np.uint8)
B_transformed = np.clip(a * B + b, 0, 255).astype(np.uint8)

# Merge the transformed channels back into an image
image_transformed = cv2.merge([R_transformed, G_transformed, B_transformed])

# Visualize original and transformed images
plt.figure(figsize=(10, 5))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image (RGB)")
plt.axis("off")

# Transformed Image
plt.subplot(1, 2, 2)
plt.imshow(image_transformed)
plt.title(f"Transformed Image (a={a}, b={b})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## 6.2 HSV color space

```python
#Answer your code here.
dog = cv2.imread('./Dogg.jpg')
image_hsv = dog.copy()
image_hsv  = cv2.cvtColor(dog, cv2.COLOR_BGR2HSV)

# linear function: f(x)=ax+b
H, S, V = cv2.split(image_hsv)

# Apply linear transformations
a = 1.2  # Multiplication constant (scaling factor)
b = 100   # Addition constant (offset)

H_transformed = H
S_transformed = S
V_transformed = np.clip(a * V + b, 0, 255).astype(np.uint8)

# Merge the transformed channels back into an image
image_transformed = cv2.merge([H_transformed, S_transformed, V_transformed])
image_transformed = cv2.cvtColor(image_transformed, cv2.COLOR_HSV2RGB)
image_hsv = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2RGB)

# Visualize original and transformed images
plt.figure(figsize=(10, 5))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_hsv)
plt.title("Original Image (RGB)")
plt.axis("off")

# Transformed Image
plt.subplot(1, 2, 2)
plt.imshow(image_transformed)
plt.title(f"Transformed Image (a={a}, b={b})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## 6.3 YCbCr color space

```python
#Answer your code here.
dog = cv2.imread('./Dogg.jpg')
image_ycrcb = dog.copy()
image_ycrcb  = cv2.cvtColor(image_ycrcb, cv2.COLOR_BGR2YCrCb)

# linear function: f(x)=ax+b
Y, Cb, Cr = cv2.split(image_ycrcb)

# Apply linear transformations
a = 1.2  # Multiplication constant (scaling factor)
b = 50   # Addition constant (offset)

Y_transformed = np.clip(a * Y + b, 0, 255).astype(np.uint8)
Cb_transformed = Cb
Cr_transformed = Cr

# Merge the transformed channels back into an image
image_transformed = cv2.merge([Y_transformed, Cb_transformed, Cr_transformed])
image_transformed = cv2.cvtColor(image_transformed, cv2.COLOR_YCrCb2RGB)
image_ycrcb = cv2.cvtColor(image_hsv, cv2.COLOR_YCrCb2RGB)

# Visualize original and transformed images
plt.figure(figsize=(10, 5))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_hsv)
plt.title("Original Image (RGB)")
plt.axis("off")

# Transformed Image
plt.subplot(1, 2, 2)
plt.imshow(image_transformed)
plt.title(f"Transformed Image (a={a}, b={b})")
plt.axis("off")

plt.tight_layout()
plt.show()
```
