```python
import numpy as np
from PIL import Image
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
```

```python
import cv2
import matplotlib.pyplot as plt
img_MtEverest= cv.imread('/content/Mount_Everest.jpg')
plt.imshow(img_MtEverest[:,:,::-1])
plt.axis("off")
plt.show()
```

```python
#Add your code here!
image_url = '/content/Mount_Everest.jpg'
image = cv2.imread(image_url)

operatedImage = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
operatedImage = np.float32(operatedImage)

dest = cv2.cornerHarris(operatedImage, 2, 5, 0.07)
dest_dilate = cv2.dilate(dest, None)
new_img = image.copy()
new_img[dest_dilate > 0.01 * dest_dilate.max()]=[0, 0, 255]

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(new_img, cv2.COLOR_BGR2RGB))
plt.title('Cornor detected Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
image_url = '/content/Mount_Everest.jpg'
image = cv2.imread(image_url)

operatedImage = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
operatedImage = np.float32(operatedImage)

dest = cv2.cornerHarris(operatedImage, 2, 5, 0.07)
dest_dilate = cv2.dilate(dest, None)
new_img_1 = image.copy()
new_img_1[dest_dilate > 0.01 * dest_dilate.max()]=[0, 0, 255]

dest = cv2.cornerHarris(operatedImage, 2, 7, 0.07)
dest_dilate = cv2.dilate(dest, None)
new_img_2 = image.copy()
new_img_2[dest_dilate > 0.01 * dest_dilate.max()]=[0, 0, 255]

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(new_img_1, cv2.COLOR_BGR2RGB))
plt.title('Cornor Detected Image blockSize=5')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(new_img_2, cv2.COLOR_BGR2RGB))
plt.title('Cornor Detected Image blockSize=7')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
img1 = cv.imread('/content/Notebook01.jpeg')
plt.imshow(img1[:,:,::-1])
plt.show()
```

```python
img2 = cv.imread('/content/Notebook02.jpeg')
plt.imshow(img2[:,:,::-1])
plt.show()
```

```python
# Answer your code here
gray = cv.cvtColor(img1,cv.COLOR_BGR2GRAY)
sift = cv.SIFT_create()
kp, des = sift.detectAndCompute(gray,None)
img1_with_keypoints = cv.drawKeypoints(gray, kp, None)

gray = cv.cvtColor(img2,cv.COLOR_BGR2GRAY)
sift = cv.SIFT_create()
kp, des = sift.detectAndCompute(gray,None)
img2_with_keypoints = cv.drawKeypoints(gray, kp, None)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img1_with_keypoints, cv2.COLOR_BGR2RGB))
plt.title('Image 1')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img2_with_keypoints, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.axis('off')

plt.tight_layout()
plt.show()
```
