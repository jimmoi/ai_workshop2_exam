# Section 3 : Reshape color space (YCrCb and HSV) to matrix size nx3 (2D) and show statistic values by using EDA tools.

3.1) Reshape color space (YCrCb and HSV) to matrix size nx3 (2D)

```python
#Answer your code here (For show result of YCrCb)
img_ycrcb = image_ycrcb.copy()

n_1 = img_ycrcb.shape[0] * img_ycrcb.shape[1]  # Calculate the total pixel amount

ycbcr_matrix = img_ycrcb.reshape((n_1, 3))  # YCrCb RESHAPE (n, 3)

df_ycrcb = pd.DataFrame(ycbcr_matrix, columns=['Y', 'Cr', 'Cb'])

print("YCrCb Statistics:")
print(df_ycrcb.head(5),'\n')
print("YCrCb Statistics:")
print(df_ycrcb.describe())
```

```python
#Answer your code here (For show result of HSV)
img_hsv = image_hsv.copy()

n_2 = img_hsv.shape[0] * img_hsv.shape[1]  # Calculate the total pixel amount

hsv_matrix = img_hsv.reshape((n_2, 3))  # HSV RESHAPE (n, 3)

df_hsv = pd.DataFrame(hsv_matrix, columns=['H', 'S', 'V'])

print("HSV Statistics:")
print(df_hsv.head(5),'\n')
print("HSV Statistics:")
print(df_hsv.describe())
```

3.2) Show plot(YCrCb and HSV) by using EDA tools.

```python
def CDF(x:DataFrame):
  count, bins_count = np.histogram(x, bins=256)
  pdf = count/ sum(count)
  cdf = np.cumsum(pdf)
  return cdf
```

```python
#Answer your code here (For show result of YCrCb)
_, bins_count = np.histogram(df_ycrcb['Y'], bins=256)

cdf_Y = CDF(df_ycrcb['Y'])
cdf_Cr = CDF(df_ycrcb['Cr'])
cdf_Cb = CDF(df_ycrcb['Cb'])

plt.plot(bins_count[1:], cdf_Y,color="red", label="Y")
plt.plot(bins_count[1:], cdf_Cr,color="green", label="Cr")
plt.plot(bins_count[1:], cdf_Cb,color="blue", label="Cb")
plt.legend()
```

```python
#Answer your code here (For show result of HSV)

cdf_H = CDF(df_hsv['H'])
cdf_S = CDF(df_hsv['S'])
cdf_V = CDF(df_hsv['V'])

plt.plot(bins_count[1:], cdf_H,color="red", label="H")
plt.plot(bins_count[1:], cdf_S,color="green", label="S")
plt.plot(bins_count[1:], cdf_V,color="blue", label="V")
plt.legend()
```

# Section 6. Make image brighter
* read image
* convert to target color space
* apply your logic
* convert back to RGB (if need)
* show the image before and after change

```python
def linear_contrast(image:np.ndarray, a=1, b=0):
  channel1, channel2, channel3 = cv2.split(image)

  c1_transformed = np.clip(a * channel1 + b, 0, 255).astype(np.uint8)
  c2_transformed = np.clip(a * channel2 + b, 0, 255).astype(np.uint8)
  c3_transformed = np.clip(a * channel3 + b, 0, 255).astype(np.uint8)

  image_transformed = cv2.merge([c1_transformed, c2_transformed, c3_transformed])

  return image_transformed
```

## 6.1 RGB color space

```python
#Answer your code here.
image_rgb = cv2.cvtColor(dog, cv2.COLOR_BGR2RGB)
a_rgb,b_rgb = 1.3,0
image_rgb_transformed = linear_contrast(image_rgb, a_rgb, b_rgb)

plt.figure(figsize=(8, 4))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image (RGB)")
plt.axis("off")

# Transformed Image
plt.subplot(1, 2, 2)
plt.imshow(image_rgb_transformed)
plt.title(f"Transformed Image (a={a_rgb}, b={b_rgb})")
plt.axis("off")

plt.tight_layout()
plt.show()
```
