### **Question 2 Easy: Implement an evaluation metric function and apply it to test the given dataset** **(2.5 points)**


- You are required to implement two evaluation metrics commonly used in image segmentation tasks: Dice Score and Intersection over Union (IoU). After implementing these metrics, apply the functions to the provided dataset and display the evaluation score for each image.

- ทำการ implement สองเมตริกการประเมินผลที่มักใช้ในงานการแยกส่วนของภาพ (image segmentation): Dice Score และ Intersection over Union (IoU) หลังจากที่ implement ฟังก์ชันเหล่านี้แล้ว ให้ใช้ฟังก์ชันเหล่านี้กับชุดข้อมูลที่ได้รับและแสดงคะแนนการประเมินผลสำหรับแต่ละภาพ

  $$
  \text{Dice} = \frac{2 \times |A \cap B|}{|A| + |B|}
  $$
  \
  $$
  \text{IoU} = \frac{\text{Area of Overlap}}{\text{Area of Union}}
  $$

```python
#https://medium.com/@nghihuynh_37300/understanding-evaluation-metrics-in-medical-image-segmentation-d289a373a3f
```

```python
import numpy as np
import matplotlib.pyplot as plt
def dice_score(pred, ground_truth):
  intersect = np.sum(pred*ground_truth)
  total_sum = np.sum(pred) + np.sum(ground_truth)
  dice = np.mean(2*intersect/total_sum)
  return round(dice, 3)

def iou_score(pred, ground_truth):
  intersect = np.sum(pred*ground_truth)
  union = np.sum(pred) + np.sum(ground_truth) - intersect
  iou = np.mean(intersect/union)
  return round(iou, 3)

# Random given images
np.random.seed(2025)
pred_images = np.random.randint(0, 2, (10, 64, 64))
gt_images = np.random.randint(0, 2, (10, 64, 64))

# Code here
for i in range(10):
  d = dice_score(pred_images[i],gt_images[i])
  iou = iou_score(pred_images[i],gt_images[i])

  print(f'image {i} have dice score ={d} and iou score ={iou}')
```

### **Question 4 Easy: Activation Function Implementation from Scratch** **(2.5 points)**

- Implement **ReLu** and **Leaky ReLU** from scratch and test with given array.
- Then explain the different between ReLU and Leaky ReLU
- Plot the output of these functions.

- ทำการ implement **ReLU** และ **Leaky ReLU** จากพื้นฐานและทดสอบกับอาร์เรย์ที่กำหนด
- จากนั้นอธิบายความแตกต่างระหว่าง ReLU และ Leaky ReLU
- วาดกราฟผลลัพธ์ของฟังก์ชันเหล่านี้"

$$ \text{ReLU}(x) = \max(0, x) $$
\
$$
\text{Leaky ReLU}(x) =
\begin{cases}
x & \text{if } x > 0 \\
\alpha \cdot x & \text{if } x \leq 0
\end{cases}
$$

```python
def relu(x):
  return np.where(x > 0,x,0)
```

```python
def leaky_relu(x, alpha=0.05):
  return np.where(x > 0,x,alpha*x)
```

```python
# Example input array
x = np.linspace(-5, 5, 100)
x_input = x.reshape(-1, 1)


# Apply ReLU and Leaky ReLU
relu_output = relu(x_input)
leaky_relu_output = leaky_relu(x_input)

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(relu_output)

plt.figure(figsize=(10, 6))
plt.plot(leaky_relu_output)

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load an image
cat = cv2.imread(r'D:\Lecture_Lab\Workshop_II\workshop_mockup\LabTestMaterial\material\cat.jpg')
cat = cv2.resize(cat,(600,600))
# Function to add Gaussian Noise
def add_gaussian_noise(image, mean=0, sigma=25):
    noise = np.random.normal(mean, sigma, image.shape) # generate noise
    noisy_image = image + noise #  add noise to the image
    noisy_image = np.clip(noisy_image, 0, 255)
    return noisy_image.astype(np.uint8)

# Function to add Salt-and-Pepper Noise
def add_salt_and_pepper_noise(image, salt_prob=0.01, pepper_prob=0.01):
    noisy_image = image.copy()
    # get image size
    total_pixels = image.size // image.shape[2]
    # calculate the number of noises
    num_salt = int(total_pixels * salt_prob)
    num_pepper = int(total_pixels * pepper_prob)

    # Add salt noises
    for _ in range(num_salt):
        x, y = np.random.randint(0, image.shape[1]), np.random.randint(0, image.shape[0])
        noisy_image[y, x] = [255, 255, 255] # white

    # Add pepper noises
    for _ in range(num_pepper):
        x, y = np.random.randint(0, image.shape[1]), np.random.randint(0, image.shape[0])
        noisy_image[y, x] = [0, 0, 0]  # black
    return noisy_image

# Plotting the images
g_noisy_img = add_gaussian_noise(cat,sigma=100)
plt.title('gaussian noise')
plt.imshow(cv2.cvtColor(g_noisy_img,cv2.COLOR_BGR2RGB))
plt.show()

sp_noisy_img = add_salt_and_pepper_noise(cat,salt_prob=0.1, pepper_prob=0.1)
plt.title('salt and pepper noise')
plt.imshow(cv2.cvtColor(sp_noisy_img,cv2.COLOR_BGR2RGB))
plt.show()
```

### **Question 1 Mid: Implement K-Fold Cross-Validation for Model Evaluation** **(5 points)**

**Objective**
  - Implement a machine learning pipeline that evaluates the performance of a Multilayer Perceptron (MLP) model using K-Fold Cross-Validation from scatch.
  - Implment basic MLP using pytorch:
    - Input layer: based on iris dataset
    - Hidden layer: with activation function
    - Output layer: with softmax activation
  - Metrics
    - Compute the accuracy for each fold.
    - Calculate the mean accuracy across all K folds.

**วัตถุประสงค์**
- พัฒนา Machine Learning Pipeline ที่ใช้ K-Fold Cross-Validation เพื่อตรวจสอบประสิทธิภาพของโมเดล Multilayer Perceptron (MLP) จากศูนย์
- สร้าง MLP พื้นฐาน โดยใช้ PyTorch:
    - Input Layer: อ้างอิงจาก Iris Dataset
    - Hidden Layer: มีฟังก์ชัน Activation Function
    - Output Layer: ใช้ Softmax Activation
- ตัวชี้วัด
    - คำนวณค่าความถูกต้อง (Accuracy) ของแต่ละ Fold
    - คำนวณค่าเฉลี่ย Accuracy จากทุก K-Folds

```python
import numpy as np
from sklearn.datasets import load_iris
import torch
import torch.nn as nn
import torch.optim as optim

iris = load_iris()
X = iris.data
y = iris.target

# # Code here
def k_fold_split(X, y, k=5, shuffle=True, random_state=29):
  rng = np.random.RandomState(random_state)
  random_samp = np.arange(len(X))
  if shuffle:
    rng.shuffle(random_samp)
  folds = np.array_split(random_samp, k)
  folds_list = []
  for i in range(k):
    val_idx = folds[i]
    X_val, y_val = X[val_idx], y[val_idx]
    train_idx = np.hstack([folds[j] for j in range(k) if j != i])
    X_train, y_train = X[train_idx], y[train_idx]
    folds_list.append((X_train,y_train,X_val,y_val))
  return folds_list
```

```python
X
```

```python
# Implement model
class MLP(nn.Module):
  def __init__(self, input_dim, hidden_dim, output_dim):
    super(MLP, self).__init__()
    self.fc1 = nn.Linear(input_dim, hidden_dim)
    self.relu = nn.ReLU()
    self.fc2 = nn.Linear(hidden_dim, output_dim)
    self.softmax = nn.Softmax(dim=1)

  def forward(self, x):
    x = self.fc1(x)
    x = self.relu(x)
    x = self.fc2(x)
    return self.softmax(x)
```

```python
K=5
folds = k_fold_split(X,y,k=K,shuffle=True,random_state=29)
model = MLP(4, 10, 3)
criterion = nn.CrossEntropyLoss()
epochs = 100
optimizer = optim.Adam(model.parameters(), lr=0.01) 

accuracy_ = []
for i,fold in enumerate(folds):
    X_train, y_train, X_val, y_val = fold

    # scaler = StandardScaler()
    # X_train = scaler.fit_transform(X_train)
    # X_test = scaler.transform(X_test)

    X_train = torch.tensor(X_train, dtype=torch.float32)
    X_val = torch.tensor(X_val, dtype=torch.float32)
    y_train = torch.tensor(y_train, dtype=torch.long)
    y_val = torch.tensor(y_val, dtype=torch.long)

    for epoch in range(epochs):
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, y_train)
        loss.backward() 
        optimizer.step()
    
        if (epoch + 1) % 30 == 0:
            print(f'Fold {i+1} Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

    with torch.no_grad():
        test_outputs = model(X_val)
        predicted = torch.argmax(test_outputs, dim=1)  # เลือก Class ที่มีค่า Probability สูงสุด
        accuracy = (predicted == y_val).float().mean()
        accuracy_.append(accuracy)
        print(f'Fold {i+1}, Accuracy: {accuracy:.4f}')
        
print(f'K-Fold Accuracy: {round(float(sum(accuracy_)/K),4)}')
```
