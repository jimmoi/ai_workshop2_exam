```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def calculate_optical_flow(frame1, frame2):
    # Convert frames to grayscale
    prev_gray = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    curr_gray = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Find good features to track
    prev_pts = cv2.goodFeaturesToTrack(prev_gray, maxCorners=100, qualityLevel=0.01, minDistance=7, blockSize=7)

    # Calculate optical flow
    curr_pts, status, err = cv2.calcOpticalFlowPyrLK(prev_gray, curr_gray, prev_pts, None)

    # Select good points
    good_prev = prev_pts[status == 1]
    good_curr = curr_pts[status == 1]

    return good_prev, good_curr
```

```python
# Generate sample frames (moving square)
frame1 = np.zeros((100, 100, 3), dtype=np.uint8)
frame2 = np.zeros((100, 100, 3), dtype=np.uint8)

frame1[20:40, 20:40] = 255
frame2[25:45, 25:45] = 255
```

```python
# Calculate optical flow
prev_pts, curr_pts = calculate_optical_flow(frame1, frame2)

# Visualize the results
plt.figure(figsize=(12, 4))

plt.subplot(131)
plt.imshow(cv2.cvtColor(frame1, cv2.COLOR_BGR2RGB))
plt.title('Frame 1')

plt.subplot(132)
plt.imshow(cv2.cvtColor(frame2, cv2.COLOR_BGR2RGB))
plt.title('Frame 2')

plt.subplot(133)
plt.imshow(cv2.cvtColor(frame1, cv2.COLOR_BGR2RGB))
for i, (prev, curr) in enumerate(zip(prev_pts, curr_pts)):
    x_prev, y_prev = prev.ravel()
    x_curr, y_curr = curr.ravel()
    plt.arrow(x_prev, y_prev, x_curr - x_prev, y_curr - y_prev,
              color='r', width=0.5, head_width=3, head_length=5)
plt.title('Optical Flow')

plt.tight_layout()
plt.show()

# Print the flow vectors
for i, (prev, curr) in enumerate(zip(prev_pts, curr_pts)):
    x_prev, y_prev = prev.ravel()
    x_curr, y_curr = curr.ravel()
    print(f"Point {i}: Movement from ({x_prev:.2f}, {y_prev:.2f}) to ({x_curr:.2f}, {y_curr:.2f})")
```

```python
def calculate_optical_flow(frame1, frame2, maxCorners=100, qualityLevel=0.01, minDistance=7, blockSize=7):
    # Convert frames to grayscale
    prev_gray = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    curr_gray = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Find good features to track
    prev_pts = cv2.goodFeaturesToTrack(prev_gray, maxCorners, qualityLevel, minDistance, blockSize)

    # Calculate optical flow
    curr_pts, status, err = cv2.calcOpticalFlowPyrLK(prev_gray, curr_gray, prev_pts, None)

    # Select good points
    good_prev = prev_pts[status == 1]
    good_curr = curr_pts[status == 1]

    return good_prev, good_curr
```

## 5.เปลี่ยนไปใช้ dense optical flow โดยใช้วิธี Farneback (cv2.calcOpticalFlowFarneback())

เปรียบเทียบผลลัพธ์กับวิธี sparse Lucas-Kanade
อภิปรายข้อดีและข้อเสียของแต่ละวิธี

```python
def calculate_dense_optical_flow(frame1, frame2):
    # Convert frames to grayscale
    prev_gray = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    curr_gray = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Calculate dense optical flow using Farneback method
    flow = cv2.calcOpticalFlowFarneback(prev_gray, curr_gray, None,
                                        pyr_scale=0.5, levels=3, winsize=15,
                                        iterations=3, poly_n=5, poly_sigma=1.2, flags=0)

    return flow  # Returns a flow field (2D vector for each pixel)
```

```python
prev_pts, curr_pts = calculate_optical_flow(frame3_1, frame3_x)
# คำนวณ Dense Optical Flow
flow = calculate_dense_optical_flow(frame3_1, frame3_x)

# แปลง Optical Flow เป็นแผนที่สี HSV
hsv = np.zeros_like(frame3_1)
hsv[..., 1] = 255  # Set saturation to maximum

# คำนวณมุมและขนาดของเวกเตอร์การเคลื่อนที่
mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])
hsv[..., 0] = np.uint8(ang * (180 / np.pi / 2)) # แปลงมุมเป็นค่า Hue (สี)
hsv[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)  # Normalize magnitude เป็น Value (ความสว่าง)

# แปลงจาก HSV กลับเป็น RGB เพื่อแสดงผล
optical_flow_rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

# Visualize the results
plt.figure(figsize=(12, 8))

plt.subplot(141)
plt.imshow(cv2.cvtColor(frame3_1, cv2.COLOR_BGR2RGB))
plt.title('Frame 1')

plt.subplot(142)
plt.imshow(cv2.cvtColor(frame3_x, cv2.COLOR_BGR2RGB))
plt.title('Frame X')

plt.subplot(143)
plt.imshow(cv2.cvtColor(frame3_1, cv2.COLOR_BGR2RGB))
for i, (prev, curr) in enumerate(zip(prev_pts, curr_pts)):
    x_prev, y_prev = prev.ravel()
    x_curr, y_curr = curr.ravel()
    plt.arrow(x_prev, y_prev, x_curr - x_prev, y_curr - y_prev,
              color='r', width=0.5, head_width=3, head_length=5)
plt.title('Optical Flow')

plt.subplot(144)
plt.imshow(optical_flow_rgb)
plt.title('Dense Optical Flow (Farneback)')


plt.tight_layout()
plt.show()
```

### เปรียบเทียบผลลัพธ์กับวิธี sparse Lucas-Kanade อภิปรายข้อดีและข้อเสียของแต่ละวิธี

**Sparse Lucas-Kanade (Optical Flow)**

**Advantages:**
High efficiency: This method calculates only key points in the image, such as areas with significant changes (corners), reducing computational load. It is suitable for tasks requiring speed, such as real-time object tracking.

High accuracy at key points: Since it focuses on key points, the calculation is highly accurate in areas with clear changes.

**Disadvantages:**
Does not cover the entire image: This method cannot detect motion in every pixel because it works only on key points.

Requires distinct key points: If the image lacks clear key points, such as smooth surfaces or low-detail areas, the results may not be effective.

**Dense Optical Flow (Farneback)**

**Advantages:**
Covers all pixels: This method calculates motion for every pixel in the image, allowing detection of changes across the entire frame.

Suitable for in-depth analysis: It is used when detailed motion information is needed across the whole image, such as segmentation or motion analysis.

**Disadvantages:**
Resource-intensive: Calculating every pixel requires more time and memory than Sparse Lucas-Kanade, making it less suitable for real-time applications.

Sensitive to noise: Since it calculates all pixels, noise or irrelevant details in the image can lead to inaccuracies.



---


**Sparse Lucas-Kanade (Optical Flow)**

**ข้อดี:**
ประสิทธิภาพสูง: วิธีนี้คำนวณเฉพาะจุดเด่น (key points) ในภาพ เช่น จุดที่มีการเปลี่ยนแปลงชัดเจน (corners) ซึ่งช่วยลดภาระในการประมวลผล ทำให้เหมาะสำหรับงานที่ต้องการความเร็ว เช่น การติดตามวัตถุแบบเรียลไทม์

ความแม่นยำในจุดเด่น: เนื่องจากเน้นเฉพาะจุดสำคัญ การคำนวณจึงมีความแม่นยำสูงในบริเวณที่มีการเปลี่ยนแปลงชัดเจน

**ข้อเสีย:**
ไม่ครอบคลุมพื้นที่ทั้งหมดของภาพ: วิธีนี้ไม่สามารถตรวจจับการเคลื่อนไหวในทุกพิกเซลได้ เพราะทำงานเฉพาะกับจุดเด่นเท่านั้น

ต้องการจุดเด่นที่ชัดเจน: หากภาพไม่มีจุดเด่นที่ชัดเจน เช่น พื้นผิวเรียบหรือไม่มีรายละเอียด การคำนวณอาจไม่ได้ผลลัพธ์ที่ดี


**Dense Optical Flow (Farneback)**

**ข้อดี:**
ครอบคลุมทุกพิกเซล: วิธีนี้คำนวณการเคลื่อนไหวในทุกพิกเซลของภาพ ทำให้สามารถตรวจจับการเปลี่ยนแปลงได้ทั่วทั้งเฟรม

เหมาะสำหรับการวิเคราะห์ภาพเชิงลึก: ใช้ในกรณีที่ต้องการข้อมูลเกี่ยวกับการเคลื่อนไหวในทุกส่วนของภาพ เช่น การทำ segmentation หรือ motion analysis

**ข้อเสีย:**
ใช้ทรัพยากรสูง: การคำนวณทุกพิกเซลต้องใช้เวลาและหน่วยความจำมากกว่า Sparse Lucas-Kanade จึงอาจไม่เหมาะสำหรับงานเรียลไทม์

ไวต่อ noise: เนื่องจากคำนวณทุกพิกเซล หากภาพมี noise หรือรายละเอียดที่ไม่เกี่ยวข้อง ผลลัพธ์อาจมีความผิดพลาดเพิ่มขึ้น
