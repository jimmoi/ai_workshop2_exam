### **Question 2 Easy: Implement an evaluation metric function and apply it to test the given dataset** **(2.5 points)**


- You are required to implement two evaluation metrics commonly used in image segmentation tasks: Dice Score and Intersection over Union (IoU). After implementing these metrics, apply the functions to the provided dataset and display the evaluation score for each image.

- ทำการ implement สองเมตริกการประเมินผลที่มักใช้ในงานการแยกส่วนของภาพ (image segmentation): Dice Score และ Intersection over Union (IoU) หลังจากที่ implement ฟังก์ชันเหล่านี้แล้ว ให้ใช้ฟังก์ชันเหล่านี้กับชุดข้อมูลที่ได้รับและแสดงคะแนนการประเมินผลสำหรับแต่ละภาพ

  $$
  \text{Dice} = \frac{2 \times |A \cap B|}{|A| + |B|}
  $$
  \
  $$
  \text{IoU} = \frac{\text{Area of Overlap}}{\text{Area of Union}}
  $$

```python
import numpy as np

def dice_score(pred, ground_truth):
  intersection = np.sum(pred * ground_truth)
  dice_coefficient = (2. * intersection) / (np.sum(pred) + np.sum(ground_truth))
  return dice_coefficient

def iou_score(pred, ground_truth):
  intersection = np.logical_and(pred, ground_truth).sum()
  union = np.logical_or(pred, ground_truth).sum()
  if union == 0:
    return 1.0
  return intersection / union

# Random given images
np.random.seed(2025)
pred_images = np.random.randint(0, 2, (10, 64, 64))
gt_images = np.random.randint(0, 2, (10, 64, 64))

# Code here
dice_coef = dice_score(pred_images,gt_images)
iou = iou_score(pred_images,gt_images)
print("Dice Coefficient:", dice_coef)
print("IoU:", iou)
```

- Implement **ReLu** and **Leaky ReLU** from scratch and test with given array.
- Then explain the different between ReLU and Leaky ReLU
- Plot the output of these functions.

- ทำการ implement **ReLU** และ **Leaky ReLU** จากพื้นฐานและทดสอบกับอาร์เรย์ที่กำหนด
- จากนั้นอธิบายความแตกต่างระหว่าง ReLU และ Leaky ReLU
- วาดกราฟผลลัพธ์ของฟังก์ชันเหล่านี้"

$$ \text{ReLU}(x) = \max(0, x) $$
\
$$
\text{Leaky ReLU}(x) =
\begin{cases}
x & \text{if } x > 0 \\
\alpha \cdot x & \text{if } x \leq 0
\end{cases}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(x):
  return np.array([max(0,i[0]) for i in x])
```

```python
def leaky_relu(x, alpha=0.05):
  out = []
  for i in x:
    if i[0] > 0:
      out.append(i[0])
    elif i[0] <= 0:
      out.append(alpha*i[0])
  return np.array(out)
```

```python
# Example input array
x = np.linspace(-5, 5, 100)
x_input = x.reshape(-1, 1)

# Apply ReLU and Leaky ReLU
relu_output = relu(x_input)
leaky_relu_output = leaky_relu(x_input)

print("Ans: ฟังก์ชัน relu เมื่อรับค่าเข้ามาเมื่อค่าน้อยกว่าหรือเท่ากับ 0 จะส่งค่า 0 กลับไป และเมื่อค่ามากกว่า 0 จะส่งค่าเดิมกลับไป\nในขณะที่ leaky relu ค่าที่น้อยกว่าหรือเท่ากับ 0 นั้นจะถูกคูณด้วย alpha ก่อนคืนค่ากลับไป")
# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(relu_output,label='relu')
plt.plot(leaky_relu_output, label='leaky relu')
plt.xlabel('sample')
plt.ylabel('value')
plt.title('relu vs leaky relu')
plt.legend()
plt.show()
```

- In this task, you will add noise to a given image using either Gaussian Noise or Salt-and-Pepper Noise, and then visualize the original image alongside the noisy image using matplotlib.
  - Used cat.jpg in folder material

- ให้ทำการเพิ่ม noise ให้กับภาพที่กำหนดโดยใช้  Gaussian หรือ Salt-and-Pepper แล้วทำการแสดงผลภาพต้นฉบับพร้อมกับภาพที่มี noise รบกวนโดยใช้ matplotlib

    - ใช้ภาพ cat.jpg จากโฟลเดอร์ material

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load an image
image = cv2.imread('cat.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image = cv2.resize(image, (224, 224))
# Function to add Gaussian Noise
def add_gaussian_noise(image, mean=0, sigma=25):
    row, col, ch = image.shape
    gauss = np.random.normal(mean, sigma, (row, col, ch)).astype(np.uint8)
    noisy_image = cv2.add(image, gauss)
    return noisy_image

# Function to add Salt-and-Pepper Noise
def add_salt_and_pepper_noise(image, salt_prob=0.01, pepper_prob=0.01):
    noisy_image = np.copy(image)
    total_pixels = image.shape[0] * image.shape[1]
    num_salt = int(total_pixels * salt_prob)
    num_pepper = int(total_pixels * pepper_prob)
    # Salt
    coords = [np.random.randint(0, i - 1, num_salt) for i in image.shape[:2]]
    noisy_image[coords[0], coords[1], :] = 255
    # Pepper
    coords = [np.random.randint(0, i - 1, num_pepper) for i in image.shape[:2]]
    noisy_image[coords[0], coords[1], :] = 0
    return noisy_image

# Add noise
gaussian_noisy_image = add_gaussian_noise(image)
salt_pepper_noisy_image = add_salt_and_pepper_noise(image)
# Plotting the images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(gaussian_noisy_image)
plt.title("Gaussian Noise")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(salt_pepper_noisy_image)
plt.title("Salt-and-Pepper Noise")
plt.axis('off')

plt.show()
```

- ในส่วนของ U-Net: มี 4 Encoder, 1 Bottleneck, 4 Decoder และตามด้วย Output Layer

\
💡 คำแนะนำ: ให้ทำ **ResidualBlock class** ให้สมบูรณ์ตามรูปภาพ และใช้ภายใน **ResidualUNet class**

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# Build Resudiual Block
class ResidualBlock(nn.Module):
  def __init__(self, in_channels, out_channels):
    super(ResidualBlock, self).__init__()
    self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1)
    self.bn1 = nn.BatchNorm2d(out_channels)
    self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1)
    self.bn2 = nn.BatchNorm2d(out_channels)

    self.skip = nn.Conv2d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else nn.Identity()

  def forward(self, x):
    identity = self.skip(x)
    x = F.relu(self.bn1(self.conv1(x)))
    x = self.bn2(self.conv2(x))
    x += identity    # Add skip connection

    return F.relu(x)
```

```python
class ResidualUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super(ResidualUNet, self).__init__()

        # Encoder part (downsampling)
        self.enc1 = ResidualBlock(in_channels, 64)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.enc2 = ResidualBlock(64, 128)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.enc3 = ResidualBlock(128, 256)
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.enc4 = ResidualBlock(256, 512)
        self.pool4 = nn.MaxPool2d(kernel_size=2, stride=2)

        # Bottleneck
        self.bottleneck = ResidualBlock(512, 1024)

        # Decoder part (upsampling)
        self.upconv4 = nn.ConvTranspose2d(1024, 512, kernel_size=2, stride=2)
        self.dec4 = ResidualBlock(512*2, 512)
        self.upconv3 = nn.ConvTranspose2d(512, 256, kernel_size=2, stride=2)
        self.dec3 = ResidualBlock(256*2, 256)
        self.upconv2 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.dec2 = ResidualBlock(128*2, 128)
        self.upconv1 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.dec1 = ResidualBlock(64*2, 64)

        # Final output layer
        self.final = nn.Conv2d(64, out_channels, kernel_size=1)

    def forward(self, x):
        # Encoder Path
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool1(enc1))
        enc3 = self.enc3(self.pool2(enc2))
        enc4 = self.enc4(self.pool3(enc3))

        # Bottleneck
        bottleneck = self.bottleneck(self.pool4(enc4))

        # Decoder Path
        up4 = self.upconv4(bottleneck)
        up4 = torch.cat((up4, enc4), dim=1)
        dec4 = self.dec4(up4)

        up3 = self.upconv3(dec4)
        up3 = torch.cat((up3, enc3), dim=1)
        dec3 = self.dec3(up3)

        up2 = self.upconv2(dec3)
        up2 = torch.cat((up2, enc2), dim=1)
        dec2 = self.dec2(up2)

        up1 = self.upconv1(dec2)
        up1 = torch.cat([up1, enc1], dim=1)
        dec1 = self.dec1(up1)

        # Final output
        x = self.final(dec1)
        return x
```

```python
import pickle
import matplotlib.pyplot as plt
import numpy as np

# Load pickle file
with open(r"C:\Users\computer\Desktop\test_workshop\LabTestMaterial\material\hard3_pickle.pkl", "rb") as f:
    data, labels = pickle.load(f)

num_images = len(data)
cols = 5
rows = (num_images // cols) + (num_images % cols > 0)
plt.figure(figsize=(15, 3 * rows))

# Loop through and plot all images
for i in range(num_images):
    plt.subplot(rows, cols, i + 1)
    plt.imshow(data[i])
    plt.title(f"Label: {labels[i]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
