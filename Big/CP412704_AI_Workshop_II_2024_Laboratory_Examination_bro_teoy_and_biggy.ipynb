### **Question 2 Easy: Implement an evaluation metric function and apply it to test the given dataset** **(2.5 points)**


- You are required to implement two evaluation metrics commonly used in image segmentation tasks: Dice Score and Intersection over Union (IoU). After implementing these metrics, apply the functions to the provided dataset and display the evaluation score for each image.

- ทำการ implement สองเมตริกการประเมินผลที่มักใช้ในงานการแยกส่วนของภาพ (image segmentation): Dice Score และ Intersection over Union (IoU) หลังจากที่ implement ฟังก์ชันเหล่านี้แล้ว ให้ใช้ฟังก์ชันเหล่านี้กับชุดข้อมูลที่ได้รับและแสดงคะแนนการประเมินผลสำหรับแต่ละภาพ

  $$
  \text{Dice} = \frac{2 \times |A \cap B|}{|A| + |B|}
  $$
  \
  $$
  \text{IoU} = \frac{\text{Area of Overlap}}{\text{Area of Union}}
  $$

```python
import numpy as np

def dice_score(pred, ground_truth, smooth=1e-6):
    total_dice = 0
    for i in range(pred.shape[0]):
        pred_flat = pred[i].flatten()
        gt_flat = ground_truth[i].flatten()

        intersection = np.sum(pred_flat * gt_flat)
        dice = (2. * intersection + smooth) / (np.sum(pred_flat) + np.sum(gt_flat) + smooth)
        print(f"Image {i+1} - Dice Score: {dice:.4f}")
        total_dice += dice
    return total_dice / pred.shape[0]

def iou_score(pred, ground_truth, smooth=1e-6):
    total_iou = 0
    for i in range(pred.shape[0]):
        pred_flat = pred[i].flatten()
        gt_flat = ground_truth[i].flatten()

        intersection = np.sum(pred_flat * gt_flat)
        union = np.sum(pred_flat) + np.sum(gt_flat) - intersection
        iou = (intersection + smooth) / (union + smooth)
        print(f"Image {i+1} - IoU Score: {iou:.4f}")
        total_iou += iou
    return total_iou / pred.shape[0]

# Random given images
np.random.seed(2025)
pred_images = np.random.randint(0, 2, (10, 64, 64))
gt_images = np.random.randint(0, 2, (10, 64, 64))

# Run evaluation
print("\nEvaluating Dice Score:")
mean_dice = dice_score(pred_images, gt_images)
print(f"\nMean Dice Score: {mean_dice:.4f}")

print("\nEvaluating IoU Score:")
mean_iou = iou_score(pred_images, gt_images)
print(f"\nMean IoU Score: {mean_iou:.4f}")
```

### **Question 4 Easy: Activation Function Implementation from Scratch** **(2.5 points)**

- Implement **ReLu** and **Leaky ReLU** from scratch and test with given array.
- Then explain the different between ReLU and Leaky ReLU
- Plot the output of these functions.

- ทำการ implement **ReLU** และ **Leaky ReLU** จากพื้นฐานและทดสอบกับอาร์เรย์ที่กำหนด
- จากนั้นอธิบายความแตกต่างระหว่าง ReLU และ Leaky ReLU
- วาดกราฟผลลัพธ์ของฟังก์ชันเหล่านี้"

$$ \text{ReLU}(x) = \max(0, x) $$
\
$$
\text{Leaky ReLU}(x) =
\begin{cases}
x & \text{if } x > 0 \\
\alpha \cdot x & \text{if } x \leq 0
\end{cases}
$$

```python
def relu(x):
    x = np.array(x)
    for i in range(x.shape[0]):
        if x[i] < 0:
            x[i] = 0 
    return x
```

```python
def leaky_relu(x, alpha=0.05):
    x = np.array(x)
    for i in range(x.shape[0]):
        if x[i] <= 0:
            x[i] = alpha*x[i]
    return x
```

```python
# Example input array
x = np.linspace(-5, 5, 100)
x_input = x.reshape(-1, 1)

# Apply ReLU and Leaky ReLU
relu_output = relu(x_input)
leaky_relu_output = leaky_relu(x_input)

# Plotting the results
plt.figure(figsize=(10, 6))

plt.plot(x, relu_output, label='ReLU', linewidth=2)
plt.plot(x, leaky_relu_output, label='Leaky ReLU', linestyle='--', linewidth=2)
plt.title('ReLU vs Leaky ReLU Activation Functions')
plt.xlabel('Input (x)')
plt.ylabel('Output')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random

def add_gaussian_noise(image, mean=0, std=25):
    gaussian = np.random.normal(mean, std, image.shape).astype(np.float32)
    noisy_image = image.astype(np.float32) + gaussian
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

def add_salt_and_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    noisy_image = image.copy()
    total_pixels = image.shape[0] * image.shape[1]

    # Salt
    num_salt = int(total_pixels * salt_prob)
    coords = [np.random.randint(0, i - 1, num_salt) for i in image.shape[:2]]
    noisy_image[coords[0], coords[1]] = 255

    # Pepper
    num_pepper = int(total_pixels * pepper_prob)
    coords = [np.random.randint(0, i - 1, num_pepper) for i in image.shape[:2]]
    noisy_image[coords[0], coords[1]] = 0

    return noisy_image


image_path = "material/cat.jpg"
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# noisy = add_gaussian_noise(image)
noisy = add_salt_and_pepper_noise(image)


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(noisy)
plt.title("Noisy Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### **Question 1 Mid: Implement K-Fold Cross-Validation for Model Evaluation** **(5 points)**

**Objective**
  - Implement a machine learning pipeline that evaluates the performance of a Multilayer Perceptron (MLP) model using K-Fold Cross-Validation from scatch.
  - Implment basic MLP using pytorch:
    - Input layer: based on iris dataset
    - Hidden layer: with activation function
    - Output layer: with softmax activation
  - Metrics
    - Compute the accuracy for each fold.
    - Calculate the mean accuracy across all K folds.

**วัตถุประสงค์**
- พัฒนา Machine Learning Pipeline ที่ใช้ K-Fold Cross-Validation เพื่อตรวจสอบประสิทธิภาพของโมเดล Multilayer Perceptron (MLP) จากศูนย์
- สร้าง MLP พื้นฐาน โดยใช้ PyTorch:
    - Input Layer: อ้างอิงจาก Iris Dataset
    - Hidden Layer: มีฟังก์ชัน Activation Function
    - Output Layer: ใช้ Softmax Activation
- ตัวชี้วัด
    - คำนวณค่าความถูกต้อง (Accuracy) ของแต่ละ Fold
    - คำนวณค่าเฉลี่ย Accuracy จากทุก K-Folds

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.utils import shuffle
import torch
import torch.nn as nn

iris = load_iris()
X = iris.data
y = iris.target

def k_fold_split(X, y, k=5, shuffle=True, random_state=29):
    if shuffle:
        X, y = shuffle(X, y, random_state=random_state)

    fold_size = len(X) // k
    folds = []
    for i in range(k):
        val_start = i * fold_size
        val_end = val_start + fold_size

        X_val = X[val_start:val_end]
        y_val = y[val_start:val_end]

        X_train = np.concatenate((X[:val_start], X[val_end:]), axis=0)
        y_train = np.concatenate((y[:val_start], y[val_end:]), axis=0)

        folds.append(((X_train, y_train), (X_val, y_val)))

    return folds
```

```python
class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# Build Resudiual Block
class ResidualBlock(nn.Module):
  def __init__(self, in_channels, out_channels):
    super(ResidualBlock, self).__init__()
    self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1)
    self.bn1 = nn.BatchNorm2d(out_channels)
    self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1)
    self.bn2 = nn.BatchNorm2d(out_channels)

    self.skip = nn.Conv2d(in_channels, out_channels, kernel_size=1) if in_channels != out_channels else nn.Identity()

  def forward(self, x):
    identity = self.skip(x)
    x = F.relu(self.bn1(self.conv1(x)))
    x = self.bn2(self.conv2(x))
    x += identity    # Add skip connection

    return F.relu(x)
```

```python
class ResidualUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super(ResidualUNet, self).__init__()

        # Encoder part (downsampling)
        self.enc1 = ResidualBlock(in_channels, 64)
        self.enc2 = ResidualBlock(64, 128)
        self.enc3 = ResidualBlock(128, 256)
        self.enc4 = ResidualBlock(256, 512)
        self.pool = nn.MaxPool2d(2, 2)
        # Bottleneck
        self.bottleneck = ResidualBlock(512, 1024)
        # Decoder part (upsampling)
        self.up4 = nn.ConvTranspose2d(1024, 512, kernel_size=2, stride=2)
        self.dec4 = ResidualBlock(1024, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, kernel_size=2, stride=2)
        self.dec3 = ResidualBlock(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.dec2 = ResidualBlock(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.dec1 = ResidualBlock(128, 64)
        # Final output layer
        self.final = nn.Conv2d(64, out_channels, kernel_size=1)

    def forward(self, x):
        # Encoder Path
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool(enc1))
        enc3 = self.enc3(self.pool(enc2))
        enc4 = self.enc4(self.pool(enc3))
        # Bottleneck
        bottleneck = self.bottleneck(self.pool(enc4))

        # Decoder Path
        up4 = self.up4(bottleneck)
        up4 = torch.cat([up4, enc4], dim=1)
        dec4 = self.dec4(up4)

        up3 = self.up3(dec4)
        up3 = torch.cat([up3, enc3], dim=1)
        dec3 = self.dec3(up3)

        up2 = self.up2(dec3)
        up2 = torch.cat([up2, enc2], dim=1)
        dec2 = self.dec2(up2)

        up1 = self.up1(dec2)
        up1 = torch.cat([up1, enc1], dim=1)
        dec1 = self.dec1(up1)

        # Final output
        output = self.final(dec1)
        return x
```

```python
def count_trainable_params(model):
    return sum(param.numel() for param in model.parameters() if param.requires_grad)
    
model = ResidualUNet(in_channels=3, out_channels=11)
print(f"Total trainable parameters: {count_trainable_params(model):,}")

x = torch.randn(1, 3, 128, 128)
out = model(x)
print(f"Output shape: {out.shape}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, random_split
import numpy as np
import matplotlib.pyplot as plt
import random
import cv2
from sklearn.model_selection import train_test_split
```

```python
# Define dataset class
class SyntheticSegmentationDataset(Dataset):
    def __init__(self, num_samples=1000, image_size=(64, 64), transform=None):
        self.image_size = image_size
        self.samples = [self.generate_synthetic_image() for _ in range(num_samples)]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        image, mask = self.samples[idx]
        return torch.tensor(image, dtype=torch.float32).unsqueeze(0), torch.tensor(mask, dtype=torch.long)

    def generate_synthetic_image(self):
        H, W = self.image_size
        image = np.zeros((H, W), dtype=np.float32)
        mask = np.zeros((H, W), dtype=np.uint8)

        valid = False
        while not valid:
            # วาดวงกลมแบบสุ่ม
            radius = random.randint(5, 10)
            cx = random.randint(radius, W - radius)
            cy = random.randint(radius, H - radius)
        
            # วาดสี่เหลี่ยมแบบสุ่ม
            rw = random.randint(8, 16)
            rh = random.randint(8, 16)
            rx = random.randint(0, W - rw)
            ry = random.randint(0, H - rh)
        
            # คำนวณ center ของวงกลมและสี่เหลี่ยม
            circle_center = np.array([cx, cy])
            rect_center = np.array([rx + rw // 2, ry + rh // 2])
            distance = np.linalg.norm(circle_center - rect_center)
        
            if distance > 15:
                valid = True

        cv2.circle(image, (cx, cy), radius, 1.0, -1)
        cv2.circle(mask, (cx, cy), radius, 1, -1)

        cv2.rectangle(image, (rx, ry), (rx + rw, ry + rh), 0.5, -1)
        cv2.rectangle(mask, (rx, ry), (rx + rw, ry + rh), 2, -1)

        return image, mask
```

```python
# Create Dataset
dataset = SyntheticSegmentationDataset(num_samples=1000, image_size=(64, 64))

train_size = int(0.7 * len(dataset))
val_size = int(0.1 * len(dataset))
test_size = len(dataset) - train_size - val_size

train_set, val_set, test_set = random_split(dataset, [train_size, val_size, test_size])

train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
val_loader = DataLoader(val_set, batch_size=32)
test_loader = DataLoader(test_set, batch_size=32)
```

```python
# Define the model
class SimpleSegNet(nn.Module):
    def __init__(self):
        super(SimpleSegNet, self).__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),  # (B, 1, 64, 64) → (B, 32, 64, 64)
            nn.ReLU(),
            nn.MaxPool2d(2),  # → (B, 32, 32, 32)

            nn.Conv2d(32, 64, kernel_size=3, padding=1),  # → (B, 64, 32, 32)
            nn.ReLU(),
            nn.MaxPool2d(2),  # → (B, 64, 16, 16)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(64, 32, kernel_size=2, stride=2),  # → (B, 32, 32, 32)
            nn.ReLU(),
            nn.ConvTranspose2d(32, 16, kernel_size=2, stride=2),  # → (B, 16, 64, 64)
            nn.ReLU(),
            nn.Conv2d(16, 3, kernel_size=1)  # Output logits: 3 channels (background, circle, rectangle)
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x  # logits (before softmax)
```

```python
# Create train, test, plot functions
def train_function(model, train_loader, val_loader, epochs=20, lr=0.001):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()
    train_losses = []
    val_losses = []
    best_model_state = None
    best_val_loss = float('inf')

    for epoch in range(epochs):
        model.train()
        running_train_loss = 0.0
        for images, masks in train_loader:
            outputs = model(images)
            loss = criterion(outputs, masks)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            running_train_loss += loss.item()

        train_losses.append(running_train_loss / len(train_loader))

        model.eval()
        running_val_loss = 0.0
        with torch.no_grad():
            for images, masks in val_loader:
                outputs = model(images)
                loss = criterion(outputs, masks)
                running_val_loss += loss.item()
        val_losses.append(running_val_loss / len(val_loader))

        print(f"Epoch {epoch+1}/{epochs} | Train Loss: {train_losses[-1]:.4f} | Val Loss: {val_losses[-1]:.4f}")

        if val_losses[-1] < best_val_loss:
            best_val_loss = val_losses[-1]
            best_model_state = model.state_dict()

    # Plot loss
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.title('Training & Validation Loss')
    plt.show()

    # Return best model
    model.load_state_dict(best_model_state)
    return model


# ---------- Test Function ----------
def test_function(model, test_loader):
    model.eval()
    dice_scores = []
    iou_scores = []

    with torch.no_grad():
        for images, masks in test_loader:
            outputs = model(images)
            preds = torch.argmax(outputs, dim=1)

            intersection = ((preds == masks) & (masks > 0)).sum().float()
            union = ((preds > 0) | (masks > 0)).sum().float()

            dice = (2 * intersection + 1e-6) / (preds[masks > 0].numel() + masks[masks > 0].numel() + 1e-6)
            iou = intersection / (union + 1e-6)

            dice_scores.append(dice.item())
            iou_scores.append(iou.item())
```

```python
# Create visualization function
def plot_function(model, dataset, num_samples=5):
    model.eval()
    plt.figure(figsize=(10, num_samples * 3))

    for i in range(num_samples):
        image, mask = dataset[i]
        with torch.no_grad():
            output = model(image.unsqueeze(0))  # เพิ่ม batch dim
        pred = torch.argmax(output.squeeze(), dim=0)  # (H, W)

        # แสดงภาพต้นฉบับ
        plt.subplot(num_samples, 3, i * 3 + 1)
        plt.imshow(image.squeeze(), cmap='gray')
        plt.title("Original Image")
        plt.axis('off')

        # แสดง ground truth mask
        plt.subplot(num_samples, 3, i * 3 + 2)
        plt.imshow(mask, cmap='jet')
        plt.title("Ground Truth Mask")
        plt.axis('off')

        # แสดง predicted mask
        plt.subplot(num_samples, 3, i * 3 + 3)
        plt.imshow(pred, cmap='jet')
        plt.title("Predicted Mask")
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
# Then train and test model
model = SimpleSegNet()
model = train_function(model, train_loader, val_loader, epochs=20)
test_function(model, test_loader)
```

```python
class SyntheticDetectionDataset(Dataset):
    def __init__(self, num_samples=1000, image_size=(64, 64), transform=None):
        self.num_samples = num_samples
        self.image_size = image_size
        self.transform = transform
        self.samples = [self.generate_synthetic_image() for _ in range(num_samples)]

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        image, bbox, label = self.samples[idx]
        image = torch.tensor(image, dtype=torch.float32).unsqueeze(0)
        bbox = torch.tensor(bbox, dtype=torch.float32)
        label = torch.tensor(label, dtype=torch.long)
        return image, bbox, label

    def generate_synthetic_image(self):
        H, W = self.image_size
        img = np.zeros((H, W), dtype=np.float32)

        obj_type = np.random.choice(["circle", "rectangle"])

        if obj_type == "circle":
            label = 0
            radius = random.randint(5, 10)
            cx = random.randint(radius, W - radius)
            cy = random.randint(radius, H - radius)
            cv2.circle(img, (cx, cy), radius, 1.0, -1)
            x_min, y_min = cx - radius, cy - radius
            x_max, y_max = cx + radius, cy + radius
        else:
            label = 1
            rw = random.randint(10, 20)
            rh = random.randint(10, 20)
            x_min = random.randint(0, W - rw)
            y_min = random.randint(0, H - rh)
            x_max = x_min + rw
            y_max = y_min + rh
            cv2.rectangle(img, (x_min, y_min), (x_max, y_max), 1.0, -1)

        bbox = [x_min, y_min, x_max, y_max]
        return img, bbox, label
```

```python
# Create Dataset
dataset = SyntheticDetectionDataset(num_samples=1000)
train_size = int(0.7 * len(dataset))
val_size = int(0.1 * len(dataset))
test_size = len(dataset) - train_size - val_size

train_set, val_set, test_set = random_split(dataset, [train_size, val_size, test_size])

train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
val_loader = DataLoader(val_set, batch_size=32)
test_loader = DataLoader(test_set, batch_size=32)
```

```python
class DetectionNet(nn.Module):
    def __init__(self):
        super(DetectionNet, self).__init__()
        self.backbone = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32), nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64), nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128), nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.flatten = nn.Flatten()
        self.dropout = nn.Dropout(0.4)

        self.shared_fc = nn.Sequential(
            nn.Linear(128, 64),
            nn.ReLU()
        )

        self.bbox_head = nn.Sequential(
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 4)
        )

        self.class_head = nn.Sequential(
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 2)
        )

    def forward(self, x):
        x = self.backbone(x)
        x = self.flatten(x)
        x = self.dropout(x)
        x = self.shared_fc(x)
        bbox = self.bbox_head(x)
        cls = self.class_head(x)
        return bbox, cls
```

```python
# Create train, test, plot functions
def train_function(model, train_loader, val_loader, epochs=500, lr=1e-3, patience=5):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    bbox_loss_fn = nn.MSELoss()
    cls_loss_fn = nn.CrossEntropyLoss()

    best_val_loss = float('inf')
    best_model_state = None
    patience_counter = 0

    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for images, bboxes, labels in train_loader:
            pred_bbox, pred_cls = model(images)
            loss_bbox = bbox_loss_fn(pred_bbox, bboxes)
            loss_cls = cls_loss_fn(pred_cls, labels)
            loss = loss_bbox + loss_cls

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for images, bboxes, labels in val_loader:
                pred_bbox, pred_cls = model(images)
                loss_bbox = bbox_loss_fn(pred_bbox, bboxes)
                loss_cls = cls_loss_fn(pred_cls, labels)
                loss = loss_bbox + loss_cls
                val_loss += loss.item()

        val_loss /= len(val_loader)
        print(f"Epoch {epoch+1}: Train Loss = {train_loss/len(train_loader):.4f}, Val Loss = {val_loss:.4f}")

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_model_state = model.state_dict()
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    model.load_state_dict(best_model_state)
    return model

def test_function(model, test_loader):
    model.eval()
    iou_scores = []
    precision_scores = []

    def compute_iou(box1, box2):
        xA = max(box1[0], box2[0])
        yA = max(box1[1], box2[1])
        xB = min(box1[2], box2[2])
        yB = min(box1[3], box2[3])

        interArea = max(0, xB - xA) * max(0, yB - yA)
        box1Area = max(0, (box1[2] - box1[0])) * max(0, (box1[3] - box1[1]))
        box2Area = max(0, (box2[2] - box2[0])) * max(0, (box2[3] - box2[1]))
        iou = interArea / float(box1Area + box2Area - interArea + 1e-6)
        return iou

    with torch.no_grad():
        for images, bboxes, labels in test_loader:
            pred_bboxes, pred_logits = model(images)
            pred_labels = torch.argmax(pred_logits, dim=1)

            for pb, pl, gt_b, gt_l in zip(pred_bboxes, pred_labels, bboxes, labels):
                iou = compute_iou(pb.numpy(), gt_b.numpy())
                precision = 1.0 if pl.item() == gt_l.item() else 0.0
                iou_scores.append(iou)
                precision_scores.append(precision)

    print(f"Test IoU: {np.mean(iou_scores):.4f}")
    print(f"Test Precision: {np.mean(precision_scores):.4f}")
```

```python
# Create visualization function
def plot_function(model, dataset, num_samples=5):
    model.eval()
    plt.figure(figsize=(12, num_samples * 3))

    class_names = ["Circle", "Rectangle"]

    for i in range(num_samples):
        image, true_bbox, true_label = dataset[i]
        with torch.no_grad():
            pred_bbox, pred_logits = model(image.unsqueeze(0))
        pred_bbox = pred_bbox.squeeze().numpy()
        pred_label = torch.argmax(pred_logits, dim=1).item()

        img_np = image.squeeze().numpy()
        img_np = (img_np * 255).astype(np.uint8)
        img_np = cv2.cvtColor(img_np, cv2.COLOR_GRAY2RGB)

        gt = true_bbox.numpy().astype(int)
        cv2.rectangle(img_np, (gt[0], gt[1]), (gt[2], gt[3]), (0, 255, 0), 2)
        cv2.putText(img_np, f"GT: {class_names[true_label]}", (gt[0], gt[1]-5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)
                    
        pb = pred_bbox.astype(int)
        cv2.rectangle(img_np, (pb[0], pb[1]), (pb[2], pb[3]), (255, 0, 0), 2)
        cv2.putText(img_np, f"Pred: {class_names[pred_label]}", (pb[0], pb[1]+12),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 1)

        plt.subplot(num_samples, 1, i + 1)
        plt.imshow(img_np)
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```
